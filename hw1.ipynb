```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams["figure.figsize"] = (15,10)

# Parameters of my Gaussian distribution
mu = 0
sigma = .1
```

```python
d = stats.norm.rvs(loc = mu, scale = sigma, size = 1000000)
x = np.linspace(mu - 5 * sigma, mu + 5 * sigma, 1000)
f = stats.norm.pdf(x, loc = mu, scale = sigma)
```

We generate $10^6$ random samples from a Gaussian (normal) distribution with mean $\mu = 0$ and standard deviation $\sigma = 0.1$. The mean sets the center of the distribution, while the standard deviation determines its width.

The probability density is

$$
f(x) = \frac{1}{\sqrt{2\pi}\sigma}\exp\left[-\frac{(x-\mu)^2}{2\sigma^2}\right].
$$

Next, let's compare the samples with this density using a 50-bin histogram. 

```python
fig, ax = plt.subplots(1, 1)

ax.hist(d, 50, density = True)
ax.plot(x, f, linewidth = 3,alpha = 0.7)

ax.tick_params(labelsize = 24)
ax.set_xlim([mu - 5 * sigma, mu + 5 * sigma])
ax.set_xlabel(r"$x$", fontsize = 16)
ax.set_ylabel(r"$f(x)$", fontsize = 16)
plt.show()
```

The first plot uses a linear vertical scale, which emphasizes the central peak of the Gaussian distribution. Most samples are concentrated near the mean $\mu = 0$, and the histogram follows the theoretical density curve. The low-density tails are harder to distinguish on this scale.

```python
fig, ax = plt.subplots(1, 1)

ax.hist(d, 50, density = True)
ax.plot(x, f, linewidth = 3, alpha = 0.7)

ax.set_yscale("log")
ax.tick_params(labelsize = 24)
ax.set_xlim([mu - 5 * sigma, mu + 5 * sigma])
ax.set_xlabel(r"$x$", fontsize = 16)
ax.set_ylabel(r"$f(x)$", fontsize = 16)
plt.show()
```

The second plot uses a logarithmic vertical scale, which makes the low-density tails easier to see. Taking the base-10 logarithm of the Gaussian probability density gives

$$
\log_{10} f(x) = -\log_{10}\left(\sqrt{2\pi}\sigma\right) - \frac{(x-\mu)^2}{2\sigma^2\ln 10}.
$$

This expression is quadratic in $x$ with a negative coefficient, so the theoretical curve appears as a downward-opening parabola on the logarithmic plot. The histogram may fluctuate more in the tails because fewer samples fall into those bins.
